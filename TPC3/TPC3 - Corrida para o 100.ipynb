### TPC3: Corrida para o 100 - Fazer logo à noite

O total começa em 0. O jogador e o computador alternam somando um número de 1 a 10 ao total. Quem atingir exatamente o número 100 vence.
- Se o computador jogar primeiro deverá vencer;
- Se o computador jogar em segundo poderá vencer se o outro jogador fizer uma aposto fora da estratégia vencedora.

Implementa o jogo com 2 vertentes: o computador joga primeiro (deverá ganhar sempre), e o computador joga em segundo lugar (poderá ganhar ou não dependendo das jogadas do outro).

```python
def jogo_computador_comeca():
    total = 0 
    print("\n--- Modo 1: Computador Começa ---") 
    print("O total começa em 0.") 

    
    computador = 1 
    total = total + computador 
    print("Computador joga:", computador, "→ total =", total) 

    
    alvos = [12, 23, 34, 45, 56, 67, 78, 89, 100] 

    while total < 100: 
        jogador = int(input("\nEscolhe um número de 1 a 10: ")) 
        while jogador < 1 or jogador > 10: 
            jogador = int(input("Inválido! Escolhe um número de 1 a 10: ")) 

        total = total + jogador 
        print("Jogador joga:", jogador, "→ total =", total)

        
        if total >= 100: 
            print("Ultrapassaste ou chegaste a 100! O computador já não precisa de jogar.") 
            return 

       
        for alvo in alvos:
            if alvo > total: 
                proximo_alvo = alvo
                break

        computador = proximo_alvo - total 
        total = total + computador 
        print("Computador joga:", computador, "→ total =", total) 

        if total == 100:
            print("\nO Computador chegou aos 100 e ganhou!") 
            return 


def jogo_jogador_comeca(): 
    total = 0 
    print("\n--- Modo 2: Tu Começas ---") 
    print("O total começa em 0.") 

    
    alvos = [1, 11, 22, 33, 44, 55, 66, 77, 88, 99, 100] 

    while total < 100: 
        jogador = int(input("\nEscolhe um número de 1 a 10: "))
        while jogador < 1 or jogador > 10:
            jogador = int(input("Inválido! Escolhe um número de 1 a 10: "))

        total += jogador
        print("Jogador joga:", jogador, "→ total =", total)

        if total == 100:
            print("\nParabéns! Chegaste aos 100 e ganhaste!")
            return
        elif total > 100:
            print("\nPassaste dos 100! Perdeste o jogo.")
            return

        
        computador = 11 - jogador 

        
        if total + computador > 100: 
            computador = 100 - total 

        total += computador 
        print("Computador joga:", computador, "→ total =", total) 

        if total == 100:
            print("\nO Computador chegou aos 100 e ganhou!")
            return


def menu_principal():
    print("=================================")
    print("      CORRIDA PARA O 100         ")
    print("=================================")
    print("1 - Computador começa (Computador ganha sempre)")
    print("2 - Jogador começa (Tu ganhas se jogares 1 e mantiveres a estratégia)")
    print("0 - Sair")

    opcao = input("\nEscolhe uma opção (0, 1, 2): ")

    if opcao == "1":
        jogo_computador_comeca()
    elif opcao == "2":
        jogo_jogador_comeca()
    elif opcao == "0":
        print("A sair do jogo...")
    else:
        print("Opção inválida. Tenta outra vez.")
        menu_principal()



menu_principal()
```
